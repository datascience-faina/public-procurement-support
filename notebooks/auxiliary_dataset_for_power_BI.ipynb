# Auxiliary – Importing the dataset into Power BI

### Purpose
Convert the dataset from pickle format (.pkl) into CSV format so it can be imported into Power BI for dashboard creation.

### Steps
- Load the original dataset
- Filter the data for Germany
- Convert the resulting dataset into CSV format
- Save the CSV file for use in Power BI

```python
# ---------------------------------------------------------
# Import moduls
# ---------------------------------------------------------

# import standard modules
import pandas as pd
import numpy as np
from pathlib import Path

import sys
from pathlib import Path


# show all columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
# ---------------------------------------------------------
# Load scripts
# ----------------------------------------------------------

%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

# found min directory
PROJECT_ROOT = Path("..").resolve()

# maindirectory sys.path
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# import function from script
from my_scripts.eda import (filter_germany)
```

```python
# ---------------------------------------------------------
# Import dataset
# ---------------------------------------------------------

df = pd.read_pickle("../data/dataset_nlp.pkl")


# Germany dataset
df_de = filter_germany(df)
```

```python
# removing unnecessary columns

necessary = ["YEAR", "ISO_COUNTRY_CODE", "TYPE_OF_CONTRACT", "TOP_TYPE", "AWARD_VALUE_EURO", "CRIT_PRICE_WEIGHT", "NUMBER_OFFERS", "AWARD_QUARTER",
               "DAYS_TO_AWARD", "CPV_CATEGORY", "CAE_TYPE_CATEGORY", "TOPIC_ID", "TOPIC_NAME"]

df_de = df_de[necessary].copy()
```
