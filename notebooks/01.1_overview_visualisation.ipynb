```python
numeric_cols = ["VALUE_EURO", "AWARD_VALUE_EURO", "NUMBER_OFFERS", "LOTS_NUMBER"]

for col in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], bins=50, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

--------------------------------------------
### Boxplots

--------------------------------------------

```python
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()

# Boxplot by contract type
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="TYPE_OF_CONTRACT", y="VALUE_EURO")
plt.xticks(rotation=45)
plt.title("VALUE_EURO by TYPE_OF_CONTRACT")
plt.show()
```

--------------------------------------------

### Correlation analysis

--------------------------------

```python
plt.figure(figsize=(12,10))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

--------------------------------------------

### Scatterplots

--------------------------------------------

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="NUMBER_OFFERS", y="VALUE_EURO", alpha=0.3)
plt.title("VALUE_EURO vs NUMBER_OFFERS")
plt.show()

plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="LOTS_NUMBER", y="AWARD_VALUE_EURO", alpha=0.3)
plt.title("AWARD_VALUE_EURO vs LOTS_NUMBER")
plt.show()
```

--------------------------------------------

### Categorical distributions

--------------------------------------------

```python
plt.figure(figsize=(10,4))
df["TYPE_OF_CONTRACT"].value_counts().plot(kind="bar")
plt.title("Distribution of TYPE_OF_CONTRACT")
plt.show()

plt.figure(figsize=(12,4))
df["ISO_COUNTRY_CODE"].value_counts().plot(kind="bar")
plt.title("Distribution of ISO_COUNTRY_CODE")
plt.show()

# CPV section
df["CPV_SECTION"] = df["MAIN_CPV_CODE_GPA"].astype(str).str[:2]

plt.figure(figsize=(12,4))
df["CPV_SECTION"].value_counts().plot(kind="bar")
plt.title("Distribution of CPV_SECTION")
plt.show()
```

--------------------------------------------

### Trends over time

--------------------------------------------


```python

plt.figure(figsize=(8,4))
df.groupby("YEAR")["VALUE_EURO"].mean().plot()
plt.title("Average VALUE_EURO per YEAR")
plt.show()

plt.figure(figsize=(8,4))
df.groupby("YEAR")["NUMBER_OFFERS"].median().plot()
plt.title("Median NUMBER_OFFERS per YEAR")
plt.show()

plt.figure(figsize=(8,4))
df["YEAR"].value_counts().sort_index().plot(kind="bar")
plt.title("Number of Tenders per YEAR")
plt.show()
```
